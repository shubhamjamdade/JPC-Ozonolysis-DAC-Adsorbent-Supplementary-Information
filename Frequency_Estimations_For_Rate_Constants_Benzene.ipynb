```python
from ase.data.pubchem import pubchem_atoms_search, pubchem_atoms_conformer_search
from ase.build import molecule
from ase.calculators.emt import EMT
from ase.optimize import QuasiNewton
from ase.thermochemistry import IdealGasThermo
from ase.vibrations import Vibrations
import numpy as np
import os, glob
import matplotlib.pyplot as plt
import re
import math

MMEN = pubchem_atoms_search(name='N,N-Dimethylethylenediamine')
O3 = molecule('O3')
Benzene = molecule('C6H6')
T = 298.15
```

```python
## Benzene + O3 = TS

# Benzene
## Calculating Frequencies Benezene


path_OUTCAR = '/storage/home/hcoda1/5/sjamdade3/p-dsholl3-0/Ozonolysis/DFT_Calculations/Benzene_O3/Frequency/Frequency_RC/Benzene'

os.chdir(path_OUTCAR)

frequency = []

for file in glob.glob("OUTCAR"):
    f = open(file,'r')
    lines=f.readlines()
    f.close()
    for line in lines:
        row=line.split()

        if not line.strip():
             continue


        if len(row)==11 and row[1] == 'f' and row[4] == 'THz':
            frequency.append(float(row[9])*0.001)                  # convert to eV from mev


## Calculating G_Benzene

from ase.thermochemistry import HarmonicThermo
from numpy import array

os.chdir(path_OUTCAR)

potential_energy = []

for file in glob.glob("OUTCAR"):
    f = open(file,'r')
    lines=f.readlines()
    f.close()
    for line in lines:
        row=line.split()

        if not line.strip():
             continue


        if len(row)==6 and row[0] == 'free' and row[2] == 'TOTEN':
            potential_energy.append(float(row[4]))


vib_energies = frequency  

potentialenergy = potential_energy[-1]

thermo = HarmonicThermo(vib_energies=vib_energies,
                        potentialenergy=potentialenergy)

F_Benzene = thermo.get_helmholtz_energy(temperature=T)



## O3

## Calculating Frequencies O3


path_OUTCAR = '/storage/home/hcoda1/5/sjamdade3/p-dsholl3-0/Ozonolysis/DFT_Calculations/Benzene_O3/Frequency/Frequency_RC/O3'

os.chdir(path_OUTCAR)

frequency = []

for file in glob.glob("OUTCAR"):
    f = open(file,'r')
    lines=f.readlines()
    f.close()
    for line in lines:
        row=line.split()

        if not line.strip():
             continue


        if len(row)==11 and row[1] == 'f' and row[4] == 'THz':
            frequency.append(float(row[9])*0.001)                  # convert to eV from mev



# Calculating G_O3

from ase.thermochemistry import HarmonicThermo
from numpy import array

os.chdir(path_OUTCAR)

potential_energy = []

for file in glob.glob("OUTCAR"):
    f = open(file,'r')
    lines=f.readlines()
    f.close()
    for line in lines:
        row=line.split()

        if not line.strip():
             continue


        if len(row)==6 and row[0] == 'free' and row[2] == 'TOTEN':
            potential_energy.append(float(row[4]))

atoms = O3
atoms.calc = EMT()
dyn = QuasiNewton(atoms)
dyn.run(fmax=0.01)
potentialenergy = potential_energy[-1]

vib_energies = frequency 

thermo = IdealGasThermo(vib_energies=vib_energies,
                        potentialenergy=potentialenergy,
                        atoms=atoms,
                        geometry='nonlinear',
                        symmetrynumber=2, spin=0)
F_O3 = thermo.get_gibbs_energy(temperature=T, pressure=101325.)

#### Transition State #####

## Calculating Frequencies TS


path_OUTCAR = '/storage/home/hcoda1/5/sjamdade3/p-dsholl3-0/Ozonolysis/DFT_Calculations/Benzene_O3/Frequency/Frequency_TS'

os.chdir(path_OUTCAR)

frequency = []

for file in glob.glob("OUTCAR"):
    f = open(file,'r')
    lines=f.readlines()
    f.close()
    for line in lines:
        row=line.split()

        if not line.strip():
             continue


        if len(row)==11 and row[1] == 'f' and row[4] == 'THz':
            frequency.append(float(row[9])*0.001)                  # convert to eV from mev



## Calculating G_TS

from ase.thermochemistry import HarmonicThermo
from numpy import array

os.chdir(path_OUTCAR)

potential_energy = []

for file in glob.glob("OUTCAR"):
    f = open(file,'r')
    lines=f.readlines()
    f.close()
    for line in lines:
        row=line.split()

        if not line.strip():
             continue


        if len(row)==6 and row[0] == 'free' and row[2] == 'TOTEN':
            potential_energy.append(float(row[4]))


vib_energies = frequency  

potentialenergy = potential_energy[-1]

thermo = HarmonicThermo(vib_energies=vib_energies,
                        potentialenergy=potentialenergy)

F_TS = thermo.get_helmholtz_energy(temperature=T)

print(f"F_TS: {F_TS}")

#### Intermediate Product #####

## Calculating Frequencies IP


path_OUTCAR = '/storage/home/hcoda1/5/sjamdade3/p-dsholl3-0/Ozonolysis/DFT_Calculations/Benzene_O3/Frequency/Frequency_Product'

os.chdir(path_OUTCAR)

frequency = []

for file in glob.glob("OUTCAR"):
    f = open(file,'r')
    lines=f.readlines()
    f.close()
    for line in lines:
        row=line.split()

        if not line.strip():
             continue


        if len(row)==11 and row[1] == 'f' and row[4] == 'THz':
            frequency.append(float(row[9])*0.001)                  # convert to eV from mev



## Calculating G_IP

from ase.thermochemistry import HarmonicThermo
from numpy import array

os.chdir(path_OUTCAR)

potential_energy = []

for file in glob.glob("OUTCAR"):
    f = open(file,'r')
    lines=f.readlines()
    f.close()
    for line in lines:
        row=line.split()

        if not line.strip():
             continue


        if len(row)==6 and row[0] == 'free' and row[2] == 'TOTEN':
            potential_energy.append(float(row[4]))


vib_energies = frequency  

potentialenergy = potential_energy[-1]

thermo = HarmonicThermo(vib_energies=vib_energies,
                        potentialenergy=potentialenergy)

F_IP = thermo.get_helmholtz_energy(temperature=T)

print(f"F_IP: {F_IP}")


# Define the variables (example values)
F_Benzene = F_Benzene * 23   # Free energy of the MMEN (example value, in kcal/mol)
print(f"F_Benzene: {F_Benzene}")
F_O3 = F_O3 * 23  # Free energy of the O3 (example value, in kcal/mol)
print(f"F_O3: {F_O3}")
F_TS = F_TS * 23  # Free energy of the transition state (example value, in kcal/mol)
print(f"F_TS: {F_TS}")
F_IP = F_IP * 23  # Free energy of the transition state (example value, in kcal/mol)
print(f"F_IP: {F_IP}")
```

```python
# Define the pre-factor and other constants
Kb = 1.380649e-23    # Boltzmann's constant, J/K
R = 8.314  # Universal gas constant in J/(mol*K)
h = 6.62607015e-34  # Planck's constant, J·s


# Calculate the rate constant

G_PRC = F_O3 + F_Benzene
diff_kcal = F_TS - F_O3 - F_Benzene
delta_H = F_IP - F_O3 - F_Benzene

print(f" Delta_G: {diff_kcal}")
print(f" Delta_H: {delta_H}")
print(f" G_PRC: {G_PRC}")

Rate_Constant = (Kb*T/h) * math.exp(-(F_TS - F_O3 - F_Benzene)*4200/ (R * T))

# Formatting the rate constant in scientific notation
Rate_Constant_sci = "{:.2e}".format(Rate_Constant)

# Print the result
print(f"Rate Constant k(T) at {T} K: {Rate_Constant_sci}")
```

```python
Converted_Rate_Constant = Rate_Constant*R*T*10**6/(101325*6.022*10**23)

# Formatting the rate constant in scientific notation
print(f"Rate Constant k₉(T) at {T} K: {Converted_Rate_Constant}")
```

```python
#MMEN
import pandas as pd
import matplotlib.pyplot as plt

# Read the Excel sheet
file_path = 'Reaction_Energetics.xlsx'
sheet_name = 'python_plots'  # Change this to your sheet name if necessary
df = pd.read_excel(file_path, sheet_name=sheet_name)

# Select specific rows and columns
# Example: Selecting rows 2 to 8 and columns 'A' and 'D'
start_row = 1
end_row = 8
columns_to_plot = ['A', 'D']

# Extract the specific rows and columns
data_to_plot = df.loc[start_row:end_row, columns_to_plot]

# Extract X-axis values from column 'A'
x_values = data_to_plot['A']

# Extract Y-axis values from column 'D' (or adjust as per your requirement)
y_values = data_to_plot['D']

# Plot the data
plt.figure(figsize=(10, 8))
plt.plot(x_values, y_values, label='D', linewidth=2.5, color='blue')

# Remove top and right spines
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Set the y-axis lower limit to -2
plt.ylim(-4, None)

# Manually specify y-ticks as strings
plt.yticks([-4, -2, 0, 2, 4, 6, 8, 10], ['-50', '-25','0', '2', '4', '6', '8', '10'], fontsize=20)

# Increase font size for labels and ticks
plt.xlabel('', fontsize=14)  # Add appropriate X-axis label if needed
plt.ylabel('Energy (kcal/mol)', fontsize=28)
plt.xlabel('Image Number', fontsize=20)
plt.title('', fontsize=20)  # Add appropriate title if needed
plt.xticks(fontsize=20)

# Show the plot
plt.show()
```
